```python
import numpy as np

data = np.random.random(size= 20)
data
```

```python
data.mean()
```

```python
data.var()
```

```python
import altair as alt
import pandas as pd
import numpy as np
import scipy.stats as stats

df_data = pd.DataFrame({"valores" : data})

x_teorico = np.linspace(0, data.max() + 1, 100)
y_teorico = stats.expon.pdf(x_teorico, scale=1)
df_teorico = pd. DataFrame ({"x": x_teorico, "y" : y_teorico})


histogram = alt.Chart(df_data).mark_bar(opacity=0.6, color="steelblue"). encode(
alt.X("valores:Q", bin=alt.Bin(maxbins=10), title="Valores"),
alt.Y("Count():Q", stack=None, title="Densidad / Frecuencia").impute(value=0)
)

linea_teorica = alt.Chart(df_teorico).mark_line(color="red", size=2).encode(
    x="X:Q",
    y="Y:Q",
)
```
