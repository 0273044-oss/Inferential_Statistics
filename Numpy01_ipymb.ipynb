### Operaciones vectoriales

Sean $(x_{1}, x_{2}, \dots, x_{n})$ y $(y_{1},y_{2}, \dots, y_{n})$ dos vectores. Entonces, podemos sumar estos dos vectores:
$$(x_{1}, x_{2}, \dots, x_{n})+ (y_{1},y_{2}, \dots, y_{n}) =(x_{1}+y_{1}, x_{2}+y_{2}, \dots, x_{n}+y_{n}),$$
y podemos multiplicar un vector por una constante $\alpha$:
$$a(x_{1}, x_{2}, \dots, x_{n})=(a x_{1}, a x_{2},\dots, a x_{n}).$$

```python
L1 + L2 # Concatenar Listas.
```

```python
# Podemos concatenar listas de distintos tamaños.

L3 = [1]

L4 = [100, 101]

L3+L4
```

## Arreglos

Un arreglo es una matriz en donde cada uno de sus elementos son numeros.

Meta: convertir las listas ya creadas a vectores (arrays).

```python
import numpy as np
```

```python
L1_arreglo = np.array(L1)

L2_arreglo = np.array(L2)

L3_arreglo = np.array(L3)

L4_arreglo = np.array(L4)
```
